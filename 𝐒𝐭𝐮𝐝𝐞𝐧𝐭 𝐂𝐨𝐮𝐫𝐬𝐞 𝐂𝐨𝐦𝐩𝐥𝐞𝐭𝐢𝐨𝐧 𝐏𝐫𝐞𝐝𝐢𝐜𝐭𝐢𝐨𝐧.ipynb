## **Outlier Analysis**

```python
Key_Numerical_Features = ['Age', 'Time_Spent_Hours', 'Satisfaction_Rating', 'Payment_Amount', 'Support_Tickets_Raised', 'Login_Frequency', 'Video_Completion_Rate', 'Quiz_Score_Avg', 'Project_Grade', 'Progress_Percentage']

print("\nOutlier Quantification using IQR Method:")
for col in Key_Numerical_Features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    Outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    Num_Outliers = len(Outliers)
    Percentage_Outliers = (Num_Outliers / len(df)) * 100

    print(f"\nFeature: {col}")
    print(f"  Q1: {Q1:.2f}")
    print(f"  Q3: {Q3:.2f}")
    print(f"  IQR: {IQR:.2f}")
    print(f"  Lower Bound: {lower_bound:.2f}")
    print(f"  Upper Bound: {upper_bound:.2f}")
    print(f"  Number of Outliers: {Num_Outliers}")
    print(f"  Percentage of Outliers: {Percentage_Outliers:.2f}%")
```

```python
plt.figure(figsize=(18, 15))
for i, col in enumerate(Key_Numerical_Features):
    plt.subplot(4, 3, i + 1) 
    sns.boxplot(y=df[col])
    plt.title(f'Box Plot of {col} (Outliers)')
    plt.ylabel(col)
    plt.tight_layout()
plt.suptitle('Outlier Visualization for Key Numerical Features', y=1.02, fontsize=20)
plt.show()
```

## **Feature Engineering for Completion Prediction**

```python
df['Engagement_Score'] = df['Login_Frequency'] * df['Average_Session_Duration_Min']

df['Learning_Effectiveness'] = df['Video_Completion_Rate'] * df['Quiz_Score_Avg']


df['Assignment_Success_Rate'] = df['Assignments_Submitted'] / (df['Assignments_Submitted'] + df['Assignments_Missed'])
df['Assignment_Success_Rate'] = df['Assignment_Success_Rate'].fillna(0)

print("New features 'Engagement_Score', 'Learning_Effectiveness', and 'Assignment_Success_Rate' created.")
df.head()
```

## **Machine Learning**

```python
y = df['Completed']
X = df.drop('Completed', axis=1)

Categorical_Cols_To_Encode = [col for col in X.select_dtypes(include=['object', 'category']).columns if col not in ['Student_ID', 'Name', 'Course_ID']]
Numerical_Cols_To_Scale = X.select_dtypes(include=['int64', 'float64']).columns

print("Target variable 'y' created.")
print("Features 'X' created (excluding 'Completed').")
print("Categorical columns identified for encoding:", Categorical_Cols_To_Encode)
print("Numerical columns identified for scaling:", Numerical_Cols_To_Scale.tolist())
```

```python
from sklearn.preprocessing import StandardScaler


X_encoded = pd.get_dummies(X[Categorical_Cols_To_Encode], drop_first=True)
print("Categorical columns one-hot encoded.")


Scaler = StandardScaler()
X_scaled = Scaler.fit_transform(X[Numerical_Cols_To_Scale])
X_scaled = pd.DataFrame(X_scaled, columns=Numerical_Cols_To_Scale, index=X.index)
print("Numerical columns scaled.")


X_processed = pd.concat([X_scaled, X_encoded], axis=1)
print("Scaled numerical features and one-hot encoded categorical features combined.")
print("Shape of X_processed:", X_processed.shape)
X_processed.head()
```

```python
from sklearn.model_selection import train_test_split


y_encoded = y.map({'Not Completed': 0, 'Completed': 1})

X_train, X_test, y_train, y_test = train_test_split(X_processed, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded)

print("Data split into training and testing sets.")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

## **Train and Evaluate Classification Models**


```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, f1_score

print("Required libraries imported successfully.")
```

```python
def train_evaluate_model(model, X_train, y_train, X_test, y_test, model_name):
    print(f"\n--- Evaluating {model_name} ---")
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    print(classification_report(y_test, y_pred))
    f1 = f1_score(y_test, y_pred, pos_label=1)
    return f1

print("Train_Evaluate_Model Function Defined.")
```

```python
Logistic_Regression_Model = LogisticRegression(random_state=42, solver='liblinear')
f1_lr = train_evaluate_model(Logistic_Regression_Model, X_train, y_train, X_test, y_test, "Logistic Regression")

print(f"Logistic Regression F1-score: {f1_lr:.4f}")
```

```python
Decision_Tree_Model = DecisionTreeClassifier(random_state=42)
f1_dt = train_evaluate_model(Decision_Tree_Model, X_train, y_train, X_test, y_test, "Decision Tree Classifier")

print(f"Decision Tree Classifier F1-score: {f1_dt:.4f}")
```

```python
Random_Forest_Model = RandomForestClassifier(random_state=42, n_estimators=100)
f1_rf = train_evaluate_model(Random_Forest_Model, X_train, y_train, X_test, y_test, "Random Forest Classifier")

print(f"Random Forest Classifier F1-score: {f1_rf:.4f}")
```

```python
print("\n--- Model Performance Comparison (F1-score for positive class) ---")
print(f"Logistic Regression F1-score: {f1_lr:.4f}")
print(f"Decision Tree Classifier F1-score: {f1_dt:.4f}")
print(f"Random Forest Classifier F1-score: {f1_rf:.4f}")


best_f1 = max(f1_lr, f1_dt, f1_rf)

if best_f1 == f1_lr:
    best_model_name = "Logistic Regression"
elif best_f1 == f1_dt:
    best_model_name = "Decision Tree Classifier"
else:
    best_model_name = "Random Forest Classifier"

print(f"\nThe best performing model based on F1-score is: {best_model_name} with an F1-score of {best_f1:.4f}")
```

## **Explanation of Model Performance**

From the evaluation, the **Logistic Regression** model emerged as the best-performing model with an F1-score of 0.5921 for the positive class ('Completed'). This indicates that, for this specific dataset and the features engineered, a simpler linear model was more effective at balancing precision and recall for predicting course completion compared to the other models tested.

**Why Logistic Regression Performed Best (in this instance):**

1.  **Simplicity and Interpretability:** Logistic Regression is a linear model that estimates the probability of a binary outcome. Its simplicity can be an advantage when the relationship between features and the target variable is largely linear or when the noise in the data is significant enough to hinder more complex models.
2.  **Robustness to Outliers (with scaling):** While Logistic Regression can be sensitive to outliers, the numerical features were scaled using `StandardScaler` during preprocessing. This likely helped in mitigating the negative impact of outliers, allowing the model to find a more stable decision boundary.
3.  **Well-Separated Classes (to some extent):** The F1-score indicates that the classes ('Completed' vs. 'Not Completed') are reasonably separable by a linear boundary based on the processed features. The engineered features such as 'Engagement_Score' and 'Learning_Effectiveness' might have captured linear trends that Logistic Regression can effectively leverage.

**Addressing the Random Forest Classifier:**

The **Random Forest Classifier** achieved an F1-score of 0.5811, which is slightly lower than that of Logistic Regression. While Random Forests are powerful ensemble methods known for their ability to handle complex, non-linear relationships and reduce overfitting, there could be several reasons why it didn't outperform Logistic Regression in this particular scenario:

1.  **Data Characteristics:** The relationships within this dataset might be predominantly linear or piecewise linear, which Logistic Regression can model efficiently. Random Forests, despite their capability for complexity, might not find significantly better non-linear patterns if they don't exist prominently.
2.  **Hyperparameter Tuning:** The Random Forest was trained with `n_estimators=100` and `random_state=42`. While `n_estimators=100` is a reasonable starting point, the model's performance could potentially be improved with more extensive hyperparameter tuning (e.g., `max_depth`, `min_samples_split`, `min_samples_leaf`) using techniques like GridSearchCV or RandomizedSearchCV.
3.  **Feature Importance:** It's possible that while Random Forest implicitly performs feature selection, the specific combination of features and their interactions, as presented, didn't provide a substantial non-linear advantage over the linear decision boundary found by Logistic Regression.
4.  **Overfitting (less likely here):** Random Forests are generally robust to overfitting, but without cross-validation or more aggressive tuning, there's always a slight chance. However, given its performance is close to Logistic Regression, it's more likely a matter of the data's underlying structure favoring a simpler model for the chosen metrics.

In conclusion, while Random Forest is a strong contender, for this specific dataset and current feature set, Logistic Regression proved to be marginally more effective in predicting course completion based on the F1-score.
